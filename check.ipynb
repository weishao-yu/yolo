```python
import torch

# PyTorch 資訊
print('PyTorch 版本:', torch.__version__)
print('GPU(CUDA)支援?', torch.cuda.is_available())
print('程式能調用的GPU個數:', torch.cuda.device_count())
```

```python
# GPU 資訊
P = torch.cuda.get_device_properties(0)  # 第一個 GPU
print('GPU 的型號:', P.name)
print('GPU 的顯存大小:', round(P.total_memory / 2**30, 2), 'G')
```

```python
# cuDNN 資訊
print('支援 cuDNN 加速?', torch.backends.cudnn.enabled)
print('cuDNN 版本:', torch.backends.cudnn.version())
```
